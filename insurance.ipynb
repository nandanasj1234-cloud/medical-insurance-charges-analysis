```python
q1=insurance_df["charges"].quantile(0.25)
q3=insurance_df["charges"].quantile(0.75)
iqr=q3-q1
lower_c=q1-1.5*iqr
upper_c=q3+1.5*iqr
```

```python
print(lower_c)
print(upper_c)
```

```python
outliers=insurance_df[(insurance_df["charges"]<lower_c)|(insurance_df["charges"]>upper_c)]
```

```python
print(outliers)
print(len(outliers))
```

```python
q1=insurance_df["age"].quantile(0.25)
q3=insurance_df["age"].quantile(0.75)
iqr=q3-q1
lower_age=q1-1.5*iqr
upper_age=q3+1.5*iqr
print(lower_age)
print(upper_age)
outliers=insurance_df[(insurance_df["age"]<lower_age)|(insurance_df["age"]>upper_age)]
print(outliers)
print(len(outliers))
```

```python
q1=insurance_df["bmi"].quantile(0.25)
q3=insurance_df["bmi"].quantile(0.75)
iqr=q3-q1
lower_bmi=q1-1.5*iqr
upper_bmi=q3+1.5*iqr
print(lower_bmi)
print(upper_bmi)
outliers=insurance_df[(insurance_df["bmi"]<lower_age)|(insurance_df["bmi"]>upper_age)]
print(outliers)
print(len(outliers))
```
